```python
# Handle missing values

# For 'children': fill with 0 (assuming no children means 0)
df['children'] = df['children'].fillna(0).astype(int)

# For 'country': fill with 'Unknown'
df['country'] = df['country'].fillna('Unknown')

# For 'agent': fill with 0 (0 means no agent / direct booking)
df['agent'] = df['agent'].fillna(0).astype(int)

# For 'company': fill with 0 (0 means no company booking)
df['company'] = df['company'].fillna(0).astype(int)

print("✅ Missing values handled!")

# Verify no missing values remain
print("\nRemaining missing values:", df.isnull().sum().sum())
```

```python
# Check ADR statistics before outlier treatment
print("ADR Statistics BEFORE treatment:")
print(df['adr'].describe())

# Cap outliers at 99th percentile
upper_limit = df['adr'].quantile(0.99)
print(f"\nCapping ADR at 99th percentile: {upper_limit:.2f}")

df['adr'] = df['adr'].apply(lambda x: upper_limit if x > upper_limit else x)

# Remove negative ADR (if any)
df = df[df['adr'] > 0]

print(f"\n Outliers treated!")
print(f"Final dataset shape: {df.shape}")
```

```python
# Total stay duration (weekend nights + week nights)
df['total_stay'] = df['stays_in_weekend_nights'] + df['stays_in_week_nights']

print(" Created 'total_stay' feature")
print(df['total_stay'].describe())
```

```python
# Lead time buckets (categorical grouping)
def lead_time_bucket(days):
    if days <= 7:
        return 'Last Minute (0-7)'
    elif days <= 30:
        return 'Short (8-30)'
    elif days <= 90:
        return 'Medium (31-90)'
    elif days <= 180:
        return 'Long (91-180)'
    else:
        return 'Very Long (180+)'

df['lead_time_bucket'] = df['lead_time'].apply(lead_time_bucket)

print(" Created 'lead_time_bucket' feature")
print("\nLead time distribution:")
print(df['lead_time_bucket'].value_counts())
```

```python
# Total guests (adults + children + babies)
df['total_guests'] = df['adults'] + df['children'] + df['babies']

print(" Created 'total_guests' feature")
print("\nTotal guests statistics:")
print(df['total_guests'].describe())
```

```python

# Has weekend stay (at least 1 weekend night)
df['has_weekend'] = (df['stays_in_weekend_nights'] > 0).astype(int)

print(" Created 'has_weekend' feature")
print("\nWeekend stay distribution:")
print(df['has_weekend'].value_counts())
print(f"\nPercentage with weekend stay: {df['has_weekend'].mean() * 100:.1f}%")
```

```python

# Convert month names to numbers
month_map = {
    'January': 1, 'February': 2, 'March': 3, 'April': 4, 'May': 5,
    'June': 6, 'July': 7, 'August': 8, 'September': 9, 'October': 10,
    'November': 11, 'December': 12
}
df['arrival_month'] = df['arrival_date_month'].map(month_map)

# Is summer (June, July, August)
df['is_summer'] = df['arrival_month'].between(6, 8).astype(int)

print(" Created 'is_summer' feature")
print(f"\nSummer bookings: {df['is_summer'].sum():,} out of {len(df):,}")
print(f"Percentage summer bookings: {(df['is_summer'].mean() * 100):.1f}%")
```

```python
# Average Daily Rate per person
df['adr_per_person'] = df['adr'] / df['total_guests']
df['adr_per_person'] = df['adr_per_person'].replace([float('inf'), -float('inf')], 0).fillna(0)

print("Created 'adr_per_person' feature")
print("\nADR per person statistics:")
print(df['adr_per_person'].describe())
```

```python

# Show all newly created features
new_features = ['total_stay', 'lead_time_bucket', 'total_guests', 'has_weekend', 'is_summer', 'adr_per_person', 'arrival_month']
print(" New features created:", new_features)

# Preview the data with new features
print("\n📊 Preview of original and new features:")
display(df[['lead_time', 'lead_time_bucket', 'total_stay', 'total_guests', 'adr', 'adr_per_person', 'is_canceled']].head(10))
```
